```python
max_words_per_year = df.drop(columns=['Year']).idxmax(axis=1) 
max_counts_per_year = df.drop(columns=['Year']).max(axis=1)   

result = pd.DataFrame({
    'Year': df['Year'],
    'Most Frequent Word': max_words_per_year,
    'Frequency': max_counts_per_year
})

print(result)
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('lyrics.csv')

words = ['love', 'baby', 'yeah', 'night', 'dance', 'heart']

plt.figure(figsize=(14, 8))

for word in words:
    plt.plot(df['Year'], df[word], marker='o', label=word) 

plt.title('Word Frequency Trends in Lyrics (2005–2018)')
plt.xlabel('Year')
plt.ylabel('Frequency')
plt.legend(title='Words', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.grid(True)
plt.tight_layout()
plt.show()
```
